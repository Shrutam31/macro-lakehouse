### Fetching data  through API from the valet canada Website

```python
import requests
r = requests.get("https://www.bankofcanada.ca/valet/observations/FXUSDCAD/json",
                 params={"recent": 5}, timeout=30)
print(r.status_code)
print(r.text[:1500])
```

### Inspect the real data points

```python
import requests, json

r = requests.get("https://www.bankofcanada.ca/valet/observations/FXUSDCAD/json",
                 params={"recent": 5}, timeout=30)
data = r.json()

print(list(data.keys()))
print(json.dumps(data["observations"], indent=2))
```

### What a bad request returns

```python
bad = requests.get("https://www.bankofcanada.ca/valet/observations/NOT_A_SERIES/json", timeout=30)
print(bad.status_code, bad.text[:300])
```

### Testing Statstics Canada

```python
import requests
url = "https://www150.statcan.gc.ca/t1/wds/rest/getDataFromVectorsAndLatestNPeriods"
payload = [{"vectorId": 41690973, "latestN":3},
           {"vectorId": 2062815, "latestN":3}]

r = requests.post(url, json=payload, timeout = 30)
print(r.status_code)
print(r.text[:800])
```

```python
import requests

base = "https://www150.statcan.gc.ca/t1/wds/rest"

# 1) What are these vectors?
info = requests.post(f"{base}/getSeriesInfoFromVector",
                     json=[{"vectorId": 2062815}, {"vectorId": 41690973}], timeout=30)
print(info.status_code)
for item in info.json():
    print(item["status"], item["object"])

# 2) Latest data points, one line each
data = requests.post(f"{base}/getDataFromVectorsAndLatestNPeriods",
                     json=[{"vectorId": 2062815, "latestN": 3},
                           {"vectorId": 41690973, "latestN": 3}], timeout=30)
for item in data.json():
    obj = item["object"]
    print("vector", obj["vectorId"], "product", obj["productId"])
    for p in obj["vectorDataPoint"]:
        print("  ", p["refPer"], p["value"], p["releaseTime"], "status", p["statusCode"])
```
